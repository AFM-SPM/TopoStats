```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import pickle

from topostats.damage.classes import GrainCollection
```

```python
dir_local_pkl_cache = Path("/Users/sylviawhittleadmin/local-data/dna_damage_cache_cesium")
dir_remote_output = Path(
    "/Users/sylviawhittleadmin/Shares/xdrive/Shared/AFM_Data/dna_damage/Cs137_irradiations/20260204-analysis-getting-back-into-the-project/output"
)
assert dir_local_pkl_cache.exists(), f"Directory {dir_local_pkl_cache} does not exist"
dir_removed_grain_gallery = dir_remote_output / "removed_grains_gallery/"
dir_removed_grain_gallery.mkdir(parents=True, exist_ok=True)
graincrop_file = dir_local_pkl_cache / "analysis_results" / "grain_collection.pkl"
assert graincrop_file.exists(), f"Grain crop file {graincrop_file} does not exist"
rejected_graincrop_file = dir_local_pkl_cache / "unanalysed_grain_collection_contour_length_filtered.pkl"
assert rejected_graincrop_file.exists(), f"Rejected grain crop file {rejected_graincrop_file} does not exist"
with open(graincrop_file, "rb") as f:
    grain_collection: GrainCollection = pickle.load(f)
    assert isinstance(grain_collection, GrainCollection), f"Loaded object is not a GrainCollection, but {type(grain_collection)}"

defect_grain_statistics = pd.read_csv(dir_local_pkl_cache / "analysis_results" / "defect_grain_statistics.csv")
removed_grains_df = pd.read_csv(dir_local_pkl_cache / "bad_grains_contour_length_too_low.csv")
allowed_grains_df = pd.read_csv(dir_local_pkl_cache / "grain_statistics.csv")

print(len(removed_grains_df), "grains were removed due to contour length being too low")
print(len(allowed_grains_df), "grains were allowed to remain")
```

```python
print(f"std of contour length of removed grains: {removed_grains_df['total_contour_length'].std()}")
print(f"std of contour length of allowed grains: {allowed_grains_df['total_contour_length'].std()}")
sns.histplot(removed_grains_df["total_contour_length"], kde=True)
sns.histplot(allowed_grains_df["total_contour_length"], kde=True)
plt.show()
sns.histplot(removed_grains_df["volume"], kde=True)
sns.histplot(allowed_grains_df["volume"], kde=True)
plt.show()
```

```python
num_images_to_show = 1
show_plot = False
image_number = 0
print(removed_grains_df.columns)
for index, row in removed_grains_df.iterrows():
    grain_index = row["grain_number"]
    image_filename = row["image"]
    
    image_filename = image_filename + ".spm_all_splines.png"
    dir = row["basename"]
    # strip "../all_data/" from the dir string
    if dir.startswith("../all_data/"):
        dir = dir.replace("../all_data/", "")
    image_path_local = Path(dir) / "processed" / image_filename
    image_path_global = dir_remote_output / image_path_local
    if not image_path_global.exists():
        print(f"Warning: Image {image_path_global} does not exist, skipping")
        continue

    plt.imshow(plt.imread(image_path_global))
    plt.title(f"{image_filename}\nGrain total contour length: {row['total_contour_length']:.2f} nm")
    plt.savefig(dir_removed_grain_gallery / f"{image_filename}_contour_length_{row['total_contour_length']:.2f}.png")
    if show_plot:
        plt.show()
    else:
        plt.close()

    image_number += 1
    if image_number > num_images_to_show:
        break
```
