```python
import matplotlib.pyplot as plt

def stacked_bar_chart(
    data: dict[str, list[float]],
    labels: list[str],
    normalised: bool = False,
) -> None:
    """Create a stacked bar chart from the data."""
    fig, ax = plt.subplots()
    colours = [
        "#1f77b4",
        "#ff7f0e",
        "#2ca02c",
        "#d62728",
        "#9467bd",
        "#8c564b",
        "#e377c2",
        "#7f7f7f",
        "#bcbd22",
        "#17becf",
    ]

    if normalised:
        # normalise the data so that each bar sums to 1
        for key, values in data.items():
            total = sum(values)
            if total > 0:
                data[key] = [value / total for value in values]
    # for each key in the data, plot a bar that is stacked with each value in the list
    for i, (key, values) in enumerate(data.items()):
        print(key, values)
        # ax.bar(i, values[0], label=labels[0], color=colours[0])
        # ax.bar(i, values[1], bottom=values[0], label=labels[1], color=colours[1])
        for j, value in enumerate(values):
            colour = colours[j % len(colours)]
            bottom = sum(values[:j])  # sum of previous values for stacking
            ax.bar(i, value, bottom=bottom, label=labels[j], color=colour)

    ax.set_xticks(range(len(data)))
    ax.set_xticklabels(data.keys())
    plt.xticks(rotation=45, ha="right")
    handles, labels = ax.get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    ax.legend(by_label.values(), by_label.keys())
    plt.show()
```

```python
data: dict[str, list[float | int]] = {
    "high let mq 50": [46, 13],
    "high let mq 100": [53, 35],
    "high let te 50": [60, 0],
    "high let te 100": [80, 1],
    "low let mq 50": [56, 7],
    "low let mq 100": [46, 5],
    "low let te 50": [69, 0],
    "low let te 100": [81, 0],
}

stacked_bar_chart(data, labels=["no DSBs", "at least one DSB"])
stacked_bar_chart(data, labels=["no DSBs", "at least one DSB"], normalised=True)
```
